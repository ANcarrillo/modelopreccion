```
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd, matplotlib.pyplot as plt
from dotenv import load_dotenv
from src import config, datos
load_dotenv('../.env')
```

## 2.1 Recolección
Desde la base de datos (o cambie por `datos.cargar_desde_csv('data/archivo.csv')`).

```
df_original = datos.cargar_desde_bd()
print(df_original.shape)
df_original.head()
```

## 2.2 Variables relevantes y valores faltantes

```
df = datos.limpiar(df_original)
resumen = datos.resumen_exploratorio(df)
resumen
```

```
variables = config.VARIABLES_NUMERICAS + config.VARIABLES_CATEGORICAS
(df[variables].isna().mean() * 100).sort_values(ascending=False).plot.barh(figsize=(7, 7), title='% de valores faltantes')
plt.show()
```

```
df[config.VARIABLES_NUMERICAS].describe().T
```

## 2.3 Variables categóricas

```
for col in config.VARIABLES_CATEGORICAS:
    print(df[col].value_counts(dropna=False), '\n')
```

## 2.4 Valores atípicos (IQR)

```
datos.reporte_atipicos(df)
```

```
df[['promedio_periodo', 'asistencia_pct', 'horas_trabajo_semana', 'distancia_km']].plot.box(subplots=True, figsize=(12, 3))
plt.show()
```

## 2.5 Balance de clases
Si la clase `1` (desertores) es mucho menor, el entrenamiento aplica SMOTE.

```
df[config.OBJETIVO].value_counts().plot.bar(title='0 = no deserta, 1 = deserta')
plt.show()
```

## Relación de las variables con la deserción

```
df.groupby(config.OBJETIVO)[config.VARIABLES_NUMERICAS].mean().T
```

```
df[config.VARIABLES_NUMERICAS + [config.OBJETIVO]].corr()[config.OBJETIVO].drop(config.OBJETIVO).sort_values().plot.barh(figsize=(7, 7), title='Correlación con la deserción')
plt.show()
```
